# Exercice 2

```python
print(df.isnull().sum())
```

```python
print(df.isnull().sum().sum())
```

```python
mean_goals= df['Goals'].mean()
df['Goals'].fillna(mean_goals, inplace=True)
```

```python
median_assists = df["Assists"].median()
df["Assists"].fillna(median_assists, inplace=True)
```

```python
print(df.isnull().sum())
```

```python
print(df)
```

# Exercice 3


```python
print(df.duplicated())

# Détecter les doublons uniquement sur la colonne Player
print(df.duplicated(subset=["Player"]))
```

```python
# Afficher toutes les lignes dupliquées
print(df[df.duplicated()])

# Afficher les doublons basés sur Player
print(df[df.duplicated(subset=["Player"])])
```

```python
# Supprimer les doublons en gardant la première occurrence
df_unique = df.drop_duplicates(subset=["Player"], keep="first")

print(df_unique)
```

# exercice 4

```python
# Suppression des espaces avant/après
df["Team"] = df["Team"].str.strip()
df["Position"] = df["Position"].str.strip()
```

```python
df["Team"] = df["Team"].str.upper()
```

```python
print("Valeurs uniques Team avant nettoyage :", df["Team"].unique())
print("Valeurs uniques Position avant nettoyage :", df["Position"].unique())

# Après nettoyage
print("Valeurs uniques Team après nettoyage :", df["Team"].unique())
print("Valeurs uniques Position après nettoyage :", df["Position"].unique())
```

# Exercice 5

```python
print(df["Shots"].dtype)
```

```python
df["Shots"] = pd.to_numeric(df["Shots"], errors="coerce")
```

# Exercice 7

```python
mean_goals_by_position = df_cleaned.groupby("Position")["Goals"].mean()
print(mean_goals_by_position)
```

```python
top5_players = df_cleaned.sort_values(by="Goals", ascending=False).head(5)
print(top5_players[["Player", "Goals"]])
```

```python
df_cleaned["Goals_per_Shot"] = df_cleaned["Goals"] / df_cleaned["Shots"]
print(df_cleaned[["Player", "Goals", "Shots", "Goals_per_Shot"]].head())
```
